Función Potencial (Power Law)
Una función de la forma

$y = ax^b$

se conoce como función potencial (power function) o ley de potencia (power law).

Donde:

- $a$ es una constante.
- $b$ es el exponente de la función.

Propiedades

- Si $b > 0$, la función crece con $x$.
- Si $b < 0$, la función decrece con $x$.
- Cuanto mayor es $|b|$, más pronunciado es el cambio.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.1, 10, 500)


bs = [-1, -0.5, 0.5, 1]


fig, ax = plt.subplots(figsize=(7, 5))

for b in bs:
    y = x**b
    ax.plot(x, y, label=f"b = {b}")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title(r"Función potencial: $y = x^b$")
ax.grid(True, alpha=0.3)
ax.legend()

plt.show()
```
