## Para fatiga de alto número de ciclos $\rightarrow$ comportamiento elástico dominante

Ecuación de Basquin:

- Ley potencial entre tensión y vida


- $\sigma_a$ es la amplitud de tensión, $N_f$ el número de ciclos hasta rotura y $2N_f$ el número de **reversos** (un ciclo tiene dos reversos).

- En escala log-log es (casi) una recta.
$$\sigma_a = \sigma_f' \,(2N_f)^b$$
